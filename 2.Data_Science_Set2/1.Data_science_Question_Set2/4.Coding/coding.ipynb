```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

delays = [15, 30, 45, 20, 25, 100, 5, 60, 35, 50, 120, 80, 10, 75, 90, 200, 55, 40, 70, 150]

# create a dataframe
df = pd.DataFrame(delays, columns=['Delay'])
# calculate percentiles

percentiles = np.percentile(df['Delay'], [10, 25, 50, 75, 90])

print("Percentiles: 10th 25th 50th 75th 90th")
print(percentiles)

# interquartile range
IQR = percentiles[3] - percentiles[1]
print("Interquartile Range (IQR):", IQR)

# or else
Q1 = percentiles[1]
Q3 = percentiles[3]
IQR = Q3 - Q1
print("Interquartile Range (IQR):", IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in df['Delay'] if x < lower_bound or x > upper_bound]
print("Outliers:", outliers)
```

```python
salaries = [30000, 32000, 35000, 37000, 40000, 42000, 45000, 47000, 50000, 55000,
            60000, 62000, 65000, 67000, 70000, 72000, 75000, 80000, 85000, 90000]

df = pd.DataFrame(salaries, columns=['Salary'])
mean = df['Salary'].mean()
median = df['Salary'].median()
mode  = df['Salary'].mode()[0]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)

if mean > median:
    print("The distribution is right-skewed. Positive skew")
elif mean < median:
    print("The distribution is left-skewed. Negative skew")
else:
    print("The distribution is symmetric.")
```

```python
skewness = df['Salary'].skew()
print("Skewness:", skewness)

if skewness > 0:
    print("The distribution is right-skewed.")
elif skewness < 0:
    print("The distribution is left-skewed.")
else:
    print("The distribution is symmetric.")
```

```python
import numpy as np
import pandas as pd
data = {"medication_group": [110, 115, 108, 102, 107, 99, 111, 104, 109, 101],
        "placebo_group": [120, 125, 130, 122, 128, 119, 124, 127, 123, 126]}
df = pd.DataFrame(data)
df
```

```python
from scipy import stats
from scipy.stats import ttest_ind
df = pd.DataFrame(data)
t,p =ttest_ind(
    df['medication_group'], df['placebo_group'], equal_var=False)
print("T-statistic:", t)
print("P-value:", p)
if p < 0.05:
    print("Reject the null hypothesis: Significant difference between groups.") 
    print("The medication is effective")
else:
    print("Fail to reject the null hypothesis: No significant difference between groups.")
    print("The medication is not effective")
```
