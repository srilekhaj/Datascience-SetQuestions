```python
import pandas as pd
import matplotlib.pyplot as plt

# Create the dataframe
data = {
    "Product": ["A", "B", "A", "C", "B", "A", "C", "A", "B", "C",
                "A", "B", "C", "A", "B"],
    "Sales": [120, 90, 150, 80, 110, 130, 95, 160, 100, 85,
              140, 105, 90, 155, 115]
}

df = pd.DataFrame(data)
df
```

```python
product_frequency = df["Product"].value_counts()
product_frequency
```

```python
product_sales_sum = df.groupby("Product")["Sales"].sum()
product_sales_sum
```

```python
product_sales_sum.plot(kind="bar", color="skyblue")
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

```python
for product in df["Product"].unique():
    product_data = df[df["Product"] == product]
    plt.plot(product_data.index, product_data["Sales"], label=f"Product {product}")

plt.title("Sales Trend for Each Product")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()
plt.show()
```
