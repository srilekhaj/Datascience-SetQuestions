```python
import numpy as np
import pandas as pd
# Sample dataset
data = {
    'CustomerID': [1, 2, 3, 4, 5, 5, 6, 7, 8, 9],
    'Satisfaction': [8, 7, np.nan, 9, 6, 6, 10, 4, 5, 7]
}

# Create a DataFrame
df = pd.DataFrame(data)

s = [x for x in df['Satisfaction'] if not np.isnan(x)]
mean_satisfaction = np.mean(s)

df['Satisfaction'].fillna(mean_satisfaction, inplace=True)
print(df)
```

```python
data = {
'Group A': [110, 115, 108, 102, 107],
'Group B': [120, 125, 130, 122, 128]
}

from scipy.stats import ttest_ind
t, p = ttest_ind(data['Group A'], data['Group B'])
print(f"T-Statistic: {t:.2f}, P-Value: {p:.4f}")

if p < 0.05:
    print("Reject the null hypothesis: There is a significant difference between the groups.")
    print("The test is effective")
else:
    print("Fail to reject the null hypothesis: There is no significant difference between the groups.")
    print("The test is not effective")
```
